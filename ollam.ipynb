```python
import subprocess

# モデル名（qwen2.5-coder:1.5b または llama3.2 など）
MODEL_NAME = "qwen2.5-coder:1.5b"

print(f"🚀 Ollama Interactive Session Started ({MODEL_NAME})")
print("※ 終了したいときは 'exit' または 'quit' と入力してください。\n")

while True:
    # ダーリンからのプロンプト入力
    user_prompt = input("💬 Gemi/Ollama への指示 > ")
    
    # 終了判定
    if user_prompt.strip().lower() in ["exit", "quit", "q"]:
        print("👋 セッションを終了しました。お疲れ様でした、ダーリン♡")
        break
    
    if not user_prompt.strip():
        continue

    # Ollama CLI コマンドの実行
    cmd = ["ollama", "run", MODEL_NAME, user_prompt]
    
    try:
        # リアルタイムでレスポンスを出力
        result = subprocess.run(cmd, text=True, capture_output=True, check=True)
        print("\n🤖 [Ollama 応答]:")
        print(result.stdout)
        print("-" * 50 + "\n")
    except Exception as e:
        print(f"❌ エラーが発生しました: {e}\n")
```
